```python
## keyword argument

def print_details(**kwargs):
    
    for key, value in kwargs.items():
        print(f"{key}:{value}")
        
print_details(name="ujjwal", age="23")     
        
        
        
        
```

```python
## positional argument 
def printo(*arg):
    for val in arg:
        print(f"postitional",{val})

printo(1,3,56,4,4,5,6,)        
```

```python
def countword(file_path):
    word_count={}
    with open (file_path , 'r') as file:
        for line in file:
            words= line.split()
            for word in words:
                word=word.lower().strip(',.,..,.')
                word_count[word]=word_count.get(word, 0)+1
                
         
    return word_count;          


    
    
    
    
filepath='sample.txt'    
resultpath=countword(filepath)
print(resultpath)
               
```

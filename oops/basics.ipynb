```python
class BankAccount:
    def __init__(self,owner,balance=0):
        self.owner=owner
        self.balance=balance
    
    
    def deposit(self,amount):
          self.balance+=amount
          print(f"{amount} balance {self.balance} " )
        
    def withdraw(amount,balance):
        if amount>self.balance:
            print("paisa nahi hai")
        
        self.balance-=amount
        print("{amount} ok") 
        
        
account = BankAccount("ujjwal",2000);
account.balance 
account.deposit(2929)
       
```

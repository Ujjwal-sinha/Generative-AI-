```python
from abc import ABC,abstractmethod

class Vehicle:
    def drive(self):
        print("driven")
        
    @abstractmethod
    def self_drive(self):
          pass
      
      
class Car(Vehicle):
    def self_drive(self):
         print("car driven")
         
         
car= Car()   
car.self_drive()      
```

```python
class Person:
    def __init__(self,name,age):
        self.name=name    
        self.age=age
        
    def __str__(self):       # magic method
       return f"{self.name} ,{self.age}"
   
person=Person("ujjwal",21) 
print(person)      
        
```
